# Evaluation notebook
Loads held-out predictions and recomputes the metrics used in the README.

```python
import json
import pandas as pd
from sklearn.metrics import brier_score_loss, mean_absolute_error
pred = pd.read_csv('artifacts/eval_predictions.csv', parse_dates=['timestamp','forecast_day'])
metrics = json.load(open('artifacts/metrics.json'))
outage_mask = pred.outage == 1
print('Brier', round(brier_score_loss(pred.outage, pred.p_outage), 4))
print('MAE on true outages', round(mean_absolute_error(pred.loc[outage_mask,'duration_min'], pred.loc[outage_mask,'expected_duration_if_outage_min']), 2))
print(metrics)
pred.head()
```
